```python
# Banking System with List Comprehensions and Nested Structures

# Initial Data Structure
Bank_Data = {
    "Alice": {
        "Checking": [200, -50, 150, -25],
        "Saving": [500, -100, 150, 200],
        "Investment": [1000, -200, 300],
    },
    "Bob": {
        "Checking": [100, -20, 80, -10],
        "Saving": [1000, -300, 200],
        "Loan": [-500, -50, -100],
    },
    "Seyfan": {
        "Checking": [50, -20, 80],
        "Saving": [300, -100],
    },
    "Wania": {
        "Checking": [300, -50, 100],
        "Saving": [1000, -200, 50],
        "Credit Card": [-150, -50, -300],
    },
    "Dania": {
        "Checking": [500, -100, 200],
        "Saving": [2000, -500, 300],
    },
}

# Step 1: Add a New Customer
def add_customer(bank_data, customer_name):
    """Adds a new customer with empty lists for Checking and Saving accounts."""
    if customer_name not in bank_data:
        bank_data[customer_name] = {"Checking": [], "Saving": []}
        print(f"Customer {customer_name} added successfully.")
    else:
        print(f"Customer {customer_name} already exists.")

# Step 2: Add Transactions
def add_transaction(bank_data, customer_name, account_type, amount):
    """Adds a transaction to the specified account for a customer."""
    if customer_name in bank_data and account_type in bank_data[customer_name]:
        bank_data[customer_name][account_type].append(amount)
        print(f"Transaction of {amount} added to {customer_name}'s {account_type} account.")
    else:
        print("Invalid customer or account type.")

# Step 3: Calculate Balance
def calculate_balance(bank_data, customer_name, account_type):
    """Calculates and returns the balance for the specified account."""
    if customer_name in bank_data and account_type in bank_data[customer_name]:
        return sum([transaction for transaction in bank_data[customer_name][account_type]])
    else:
        print("Invalid customer or account type.")
        return 0

# Step 4: Filter Transactions
def filter_transactions(bank_data, customer_name, account_type, transaction_type):
    """Filters and returns deposits or withdrawals for a specified account."""
    if customer_name in bank_data and account_type in bank_data[customer_name]:
        if transaction_type == "deposit":
            return [t for t in bank_data[customer_name][account_type] if t > 0]
        elif transaction_type == "withdrawal":
            return [t for t in bank_data[customer_name][account_type] if t < 0]
        else:
            print("Invalid transaction type.")
            return []
    else:
        print("Invalid customer or account type.")
        return []

# Step 5: Generate Summary Report
def generate_summary_report(bank_data, customer_name):
    """Generates a summary report of all accounts for a customer."""
    if customer_name in bank_data:
        print(f"\nSummary Report for {customer_name}")
        for account_type, transactions in bank_data[customer_name].items():
            balance = calculate_balance(bank_data, customer_name, account_type)
            deposits = filter_transactions(bank_data, customer_name, account_type, "deposit")
            withdrawals = filter_transactions(bank_data, customer_name, account_type, "withdrawal")
            
            print(f"\n{account_type.capitalize()} Account:")
            print(f" Balance: {balance}")
            print(f" Deposits: {deposits}")
            print(f" Withdrawals: {withdrawals}")
    else:
        print(f"No data for customer {customer_name}.")

# Example Usage
if __name__ == "__main__":
    # Add a new customer
    add_customer(Bank_Data, "Semira")
    
    # Add transactions for Semira
    add_transaction(Bank_Data, "Semira", "Checking", 500)  # Deposit
    add_transaction(Bank_Data, "Semira", "Checking", -100)  # Withdrawal
    add_transaction(Bank_Data, "Semira", "Saving", 1000)  # Deposit
    add_transaction(Bank_Data, "Semira", "Saving", -200)  # Withdrawal

      # Write summary reports to a file
    with open("summary_reports.txt", "w") as file:
        file.write("Generating Summary Reports for All Customers:\n")
        file.write("=" * 50 + "\n")
        
        for customer in Bank_Data:
            file.write(f"\nSummary Report for {customer}:\n")
            file.write("=" * 50 + "\n")
            
            for account_type, transactions in Bank_Data[customer].items():
                # Calculate balance, deposits, and withdrawals
                balance = calculate_balance(Bank_Data, customer, account_type)
                deposits = filter_transactions(Bank_Data, customer, account_type, "deposit")
                withdrawals = filter_transactions(Bank_Data, customer, account_type, "withdrawal")
                
                # Write account details to the file
                file.write(f"\n{account_type.capitalize()} Account:\n")
                file.write(f" Balance: {balance}\n")
                file.write(f" Deposits: {deposits}\n")
                file.write(f" Withdrawals: {withdrawals}\n")
            
            file.write("-" * 50 + "\n")
    
    print("Summary reports have been written to 'summary_reports.txt'.")

```
